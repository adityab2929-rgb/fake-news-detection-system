```python
import pandas as pd
df=pd.read_csv('reqfiles/Fake.csv')
df.head()
```

```python
df=df.drop('date',axis=1)
```

```python
df.head()
```

```python
df['target']=1
df.head()
```

```python
df.shape
df=df.drop('subject',axis=1)
df.shape
```

```python
df2=pd.read_csv('reqfiles/True.csv')
df2=df2.drop('subject',axis=1)
df2=df2.drop('date',axis=1)
df2['target']=0
df2.shape
```

```python
#combining both the dataframes(true and Fake) into a single dataframe df
df = pd.concat([df, df2], axis=0)
df.reset_index(drop=True, inplace=True)
df = df.sample(frac=1, random_state=42).reset_index(drop=True)
df.head()
```

```python
# Basic column info and data types
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)

# Check NaN values
print("\nNaN values per column:")
print(df.isna().sum())
print(df.isnull().sum())

# Check empty strings "" and blank-space strings like " "
print("\nEmpty or blank-space values per column:")
for col in df.columns:
    if df[col].dtype == "str":
        blank_count = df[col].astype(str).str.strip().eq("").sum()
        print(f"{col}: {blank_count}")
    else:
        print(f"{col}: not a text column")
```

```python
blank_text_rows = df[df['text'].astype(str).str.strip().eq("")]
blank_text_rows.shape
```

```python
blank_text_rows.head()
blank_text_rows[blank_text_rows['target']==0].head()
```

```python
# Remove rows where text is empty or only spaces
df = df[~df["text"].astype(str).str.strip().eq("")].reset_index(drop=True)

df.shape
```

```python
#confirmation to show that blank rows are not present in df anymore
blank_text_rows = df[df['text'].astype(str).str.strip().eq("")]
blank_text_rows.shape
blank_text_rows.head()
```

```python
df.to_csv('reqfiles/AfterFE.csv',index=False)
```
